```python
missing = df.isnull().sum().sort_values(ascending= False)
missing = missing[missing > 0]
missing 
```

```python
missing_per = (df.isnull().sum()/len(df)) * 100
missing_per = missing_per[missing_per > 0]  
missing_per.sort_values(ascending= False)
```

##how are we going to handle missing values##
1. drop the column if the missing_per is > 50%
2. Fill numerical values using dif technique 
3. Fill numerical cols

```python
cols_to_drop = missing_per[missing_per > 50].index
print(f"Columns to drop: {cols_to_drop}")
df = df.drop(columns=cols_to_drop)
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns
print(f"Numerical columns: {num_cols}")
print(f"Categorical columns: {cat_cols}")
```

```python
num_missing = df[num_cols].isnull().sum().sort_values(ascending=False)
num_missing = num_missing[num_missing > 0]
num_missing
```

# Handle Missing values using dif tech {mean , median or ML algo}

using mean {when normal and no strong outliers }
using median {when skewed and there are outliers}
using Iterative method from model based 


```python
num_mean = df[num_missing.index].mean()
num_median = df[num_missing.index].median()
print(f"Numerical columns mean: {num_mean}\n")
print(f"Numerical columns median: {num_median}")
```

```python
num_df = df.select_dtypes(include=['int64', 'float64'])
num_df.head()
```

```python
cat_df = df.select_dtypes(include=['object'])
cat_df.head()
```

```python
num_missing_mean = num_df.isnull().sum().sort_values(ascending=False)
num_missing_mean = num_missing_mean[num_missing_mean > 0]
num_missing_mean
```

```python
df_mean = num_df.copy()
for col in df_mean.columns:
    df_mean[col].fillna(df_mean[col].mean(), inplace=True)
```
