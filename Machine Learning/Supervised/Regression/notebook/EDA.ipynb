```python
# classify target variable and independent variables
Y = df['SalePrice']
X = df.drop(['SalePrice'], axis=1)
```

```python
# Lets analyze the target variable 
'''Missing values
Summary statistics
Distribution
Histogram
KDE plot
Boxplot
Skewness
Kurtosis
QQ plot 
Outlier inspection
Decide whether transformation is needed'''

Is_missing = Y.isnull().sum()
print("Missing values in each column:", Is_missing )
Is_missing_per = Y.isnull().mean() * 100
print("Percentage of missing values in each column:", Is_missing_per)

def summary_statistics(Y):
    print("Summary statistics of the target variable:")
    print(f"Mean: {Y.mean()}")
    print(f"Median: {Y.median()}")
    print(f"Standard Deviation: {Y.std()}")
    print(f"Minimum: {Y.min()}")
    print(f"Maximum: {Y.max()}")
    print(f"Range: {Y.max() - Y.min()}")
    print(f"skewness: {Y.skew()}")
    print(f"kurtosis: {Y.kurtosis()}")
    print(f"Variance: {Y.var()}")
    print('='*50)

summary_statistics(Y)

# Distribution of the target variable 
# Histogram + Kde plot 
plt.figure(figsize=(12,6))
plt.subplot(1, 2, 1)
sns.histplot(Y, kde=True, bins=30)
plt.title('Histogram of SalePrice')
plt.xlabel('SalePrice')
plt.ylabel('Frequency')
plt.show() 
# Boxplot of the target variable 
plt.figure(figsize=(6 , 6))
sns.boxplot(x=Y)
plt.title('Boxplot of SalePrice')
plt.xlabel('SalePrice')
plt.show()

def outliers_check(Y):
    Q1 = Y.quantile(0.25)
    Q3 = Y.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = Y[(Y < lower_bound) | (Y > upper_bound)]
    print(f"Number of outliers: {len(outliers)}")
    print(f"Outlier values: {outliers.values}")
    return outliers
outliers = outliers_check(Y)
```

```python
# check multicollinearity using VIF (Variance Inflation Factor) and correlation matrix
corr_matrix = df[num_cols].corr()

printed = set()

for col in corr_matrix.columns:
    for row in corr_matrix.index:
        if col != row:
            pair = tuple(sorted([row, col]))
            
            if pair not in printed and abs(corr_matrix.loc[row, col]) > 0.8:
                print(f"{pair[0]} vs {pair[1]} → {corr_matrix.loc[row, col]:.2f}")
                printed.add(pair)
```

```python
# Multi variate and Bivariate Analysis
'''Mixed Analysis
Numerical vs Target
Categorical vs Target
Numerical vs Numerical
Categorical vs Categorical
Numerical vs Categorical
General Findings
Strong predictors
Weak predictors
Redundant variables
Data imbalance'''
import seaborn as sns
import matplotlib.pyplot as plt

num_cols = df.select_dtypes(include='number').columns

for col in num_cols:
    if col != "SalePrice":
        corr_matrix = df[col].corr(df['SalePrice'])
        if abs(corr_matrix) > 0.3 :
            print(f"the raltion b/n SalePrice and {col} is :{ corr_matrix}")   
```

```python
linear_features = []
nonlinear_candidates = []

for col in num_cols:
    if col == "SalePrice":
        continue

    pearson = df[col].corr(df["SalePrice"])
    spearman = df[col].corr(df["SalePrice"], method="spearman")

    difference = abs(spearman - pearson)

    if difference <= 0.1:
        linear_features.append(col)
    elif difference <= 0.2:
        print(f"{col}: Mild non-linearity")
    else:
        nonlinear_candidates.append(col)

print("\nLinear features:")
print(linear_features)

print("\nNeed further investigation:")
print(nonlinear_candidates)
```
