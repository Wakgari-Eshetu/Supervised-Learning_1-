```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
import os 
import sys
sys.path.append(os.path.join('..'))
from Config.config import raw_data_path
```

```python
df = pd.read_csv(raw_data_path)
df.head()
```

```python
target = df["SalePrice"]
values = df.drop("SalePrice", axis=1)
```

```python
num_cols = values.select_dtypes('number').columns.to_list()
cat_cols = values.select_dtypes('object').columns.to_list()
```
